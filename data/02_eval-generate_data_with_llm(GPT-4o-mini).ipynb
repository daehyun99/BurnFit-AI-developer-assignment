```python
from os import getenv
import os

import time

from dotenv import load_dotenv
from openai import OpenAI
import openai

import pandas as pd
import json
import ast
```

```python
load_dotenv()
api_key = getenv("OPENAI_API_KEY")
client = openai.OpenAI(
    api_key = api_key
)
```

# step 값 확인하기!

```python
step = 5
df = pd.read_excel(f"G:/내 드라이브/05_[공유파일]/Burnfit_dataset/eval_data{step}.xlsx")

df.head()
```

```python
outputs = []
model_name = "gpt-4o-mini"

for index, row in df.iterrows():
    try:
        user_input = row["user_input"]
        gemma3_outputs = row["gemma3_outputs"]

        prompt =f"""
        ### 가이드라인 ###
        당신은 유능한 헬스 트레이너입니다.
        현재 531 by jim wendler에 관한 질문의 답변 정확도를 평가하고, 개선된 질문과 답변을 제공하는 업무를 하고 있습니다. 
        당신의 답변은 ["평가", "개선된 질문", "개선된 답변"]의 형식을 갖추어야 합니다.
        또한, 문장 안에, 쌍따옴표(")를 사용하지마세요. ast.literal_eval() 함수를 활용하여, 파이썬 리스트 형태로 변환할 수 있어야 합니다.
        "평가"는 "적절"과 "부적절" 중에 하나의 값을 가집니다. 쌍따옴표(")로 문장을 감싸야합니다.
        "개선된 질문"은 "질문"을 개선한 문장의 형태입니다. 쌍따옴표(")로 문장을 감싸야합니다.
        "개선된 답변"은 "답변"을 개선한 문장의 형태입니다. 쌍따옴표(")로 문장을 감싸야합니다.

        ### 예시1 ###
        ["적절", "5/3/1 운동 루틴의 구성에 대하여 헬스 트레이너의 관점에서 설명해주세요.", "...(개선된 답변 작성)"]

        ### 예시2 ###
        ["부적절", "...(개선된 질문 작성)", "...(개선된 답변 작성)"]

        ### 요청 ###
        다음의 질문과 답변에 대해 평가해주세요.
        [질문]
        {user_input}
                
        [답변]
        {gemma3_outputs}

        """
        response = client.chat.completions.create(
            model=f"{model_name}",
            messages=[{"role":"user","content":[{"type":"text","text":f"{prompt}"}]}],
            temperature=0.1,
            top_p=0.8
        )
        
        result = response.choices[0].message.content
        outputs.append(result)
        print(f"index : {index} | ✅ Good")
    except Exception as e:
        outputs.append(e)
        print(f"index : {index} | ❌ ERROR : {e}")
```

```python
print(outputs[0])
print(outputs[1])
print(outputs[2])
```

```python
eval = []
question = []
answer = []

for output in outputs:
    output_ = ast.literal_eval(output)
    eval.append(output_[0])
    question.append(output_[1])
    answer.append(output_[2])
```

```python
df_result = pd.DataFrame()

df_result["eval_result"] = eval
df_result["inputs"] = question
df_result["outputs"] = answer
```

```python
df_result.head()
```

```python
df_result.to_excel(f"G:/내 드라이브/05_[공유파일]/Burnfit_dataset/eval_result{step}.xlsx", index=False)
```

```python
# Gemma3 포맷 변환 함수
def convert_to_qa_format(user_input, model_output):
    return {
        "question": user_input.strip(),
        "answer": model_output.strip()
    }

# 포맷 변환
qa_data = df_result.apply(
    lambda row: convert_to_qa_format(row["inputs"], row["outputs"]),
    axis=1
).tolist()

# JSON 저장
output_path = f"G:/내 드라이브/05_[공유파일]/Burnfit_dataset/gemma3_formatted_for_training{step+1}.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(qa_data, f, ensure_ascii=False, indent=2)

print(f"Saved to: {output_path}")
```
