```python
import pandas as pd
import random
```

# 데이터 전처리

```python
# 학습용, 평가용 데이터 로드
train_df = pd.read_excel("./dataset/train_dataset00.xlsx")
test_df = pd.read_excel("./dataset/test_dataset00.xlsx")
```

```python
# 벤치프레스의 70% 무게로 오버헤드 프레스 1RM 추가
train_df["press_1RM"] = (train_df["bench_press_1RM"] * 0.7).round(1)
test_df["press_1RM"] = (test_df["bench_press_1RM"] * 0.7).round(1)
train_df.head()
```

```python
# 운동 목표
# ENUM 값 목록
goals = [
    "최대 근력 키우기",
    "탄탄한 몸매 만들기",
    "체중 감량 성공하기",
    "운동 수행 능력 향상시키기",
    "지구력 향상시키기"
]

# 유저의 요청 샘플 목록
user_request = [
    "체중 감량도 병행 중이라 볼륨은 너무 많지 않았으면 해.",
    "하체 근력 향상에 포커스를 맞추고 싶어.",
    "요즘 허리가 안 좋아서 데드는 무리하지 않게 조절해줬으면 해.",
    "전반적인 근력과 체력을 같이 키우고 싶어.",
    "최근에 대회 준비를 시작했어. 강한 훈련이 필요해.",
    "근지구력을 높이고 싶어. 가능하면 액세서리 운동도 같이 포함해줘.",
    "운동 초보라 자세 위주로 천천히 하고 싶어.",
    "요즘 벤치가 안 늘어서 고민이야. 벤치 집중 루틴이 포함된 4주 5/3/1 프로그램 부탁해.",
    "건강 유지와 체력 향상이 목표야."
]

train_df["training_goal"] = [random.choice(goals) for _ in range(len(train_df))]
train_df["user_request"] = [random.choice(user_request) for _ in range(len(train_df))]
test_df["training_goal"] = [random.choice(goals) for _ in range(len(test_df))]
test_df["user_request"] = [random.choice(user_request) for _ in range(len(test_df))]
```

```python
print(train_df["training_goal"].value_counts())
train_df.head()
```

```python
print(test_df["training_goal"].value_counts())
test_df.head()
```

```python
user_prompts = []

for index, row in train_df.iterrows():
    gender = row["성별"]
    age = row["연령대"]
    weight = row["weight"]
    training_goal = row["training_goal"]
    squat_1RM = row["squat_1RM"]
    bench_press_1RM = row["bench_press_1RM"]
    deadlift_1RM = row["deadlift_1RM"]
    press_1RM = row["press_1RM"]


    instruction = row["user_request"] # 추후, 사용자 입력으로 확장

    input_data = {
        "성별": gender,
        "몸무게": weight,
        "squat_1RM": squat_1RM,
        "press_1RM": press_1RM,
        "bench_press_1RM": bench_press_1RM,
        "deadlift_1RM": deadlift_1RM
    }

    # user_prompt = f"""{instruction}
    
    # [input_data]
    # {input_data}
    # """
    user_prompts.append([instruction, input_data, ""])
```

```python
# ["instruction", "input", "output"] 구조의 학습용 데이터 생성
train_df = pd.DataFrame(user_prompts)

cols = ["instruction", "input", "output"]
train_df.columns = cols

train_df.head()
```

```python
user_prompts = []

for index, row in test_df.iterrows():
    gender = row["성별"]
    age = row["연령대"]
    weight = row["weight"]
    training_goal = row["training_goal"]
    squat_1RM = row["squat_1RM"]
    bench_press_1RM = row["bench_press_1RM"]
    deadlift_1RM = row["deadlift_1RM"]
    press_1RM = row["press_1RM"]


    instruction = row["user_request"] # 추후, 사용자 입력으로 확장

    input_data = {
        "성별": gender,
        "몸무게": weight,
        "squat_1RM": squat_1RM,
        "press_1RM": press_1RM,
        "bench_press_1RM": bench_press_1RM,
        "deadlift_1RM": deadlift_1RM
    }

    # user_prompt = f"""{instruction}
    
    # [input_data]
    # {input_data}
    # """
    user_prompts.append([instruction, input_data, ""])
```

```python
# ["instruction", "input", "output"] 구조의 평가용 데이터 생성
test_df = pd.DataFrame(user_prompts)

cols = ["instruction", "input", "output"]
test_df.columns = cols

test_df.head()
```

```python
# 데이터 저장

train_df.to_excel("./dataset/train_dataset01.xlsx", index=False)
test_df.to_excel("./dataset/test_dataset01.xlsx", index=False)
```
