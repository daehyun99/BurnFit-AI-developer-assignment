```python
# Common imports
import os
import jax
import jax.numpy as jnp
import optax
import treescope

# Gemma imports
from kauldron import kd
from gemma import gm
from gemma import peft

import pandas as pd
```

```python
os.environ["XLA_PYTHON_CLIENT_MEM_FRACTION"]="1.00"
```

```python
step = 6

model = gm.nn.LoRA(
    rank=4,
    model=gm.nn.Gemma3_1B(tokens="batch.input"),
)

params = gm.ckpts.load_params(f"/content/drive/MyDrive/Gemma3-lora{step}")
```

```python
tokenizer = gm.text.Gemma3Tokenizer()

sampler = gm.text.ChatSampler(
    model=model,
    params=params,
    tokenizer=tokenizer
)
```

```python
tokenizer.encode('This is an example sentence', add_bos=True)
```

## 3. 샘플 출력 확인

```python
sampler.chat("나는 웨이트 트레이닝 2년차야. 531 운동 프로그램 루틴을 작성해줘. {'성별': '남성', '몸무게': 121.05, 'squat_1RM': 100, 'press_1RM': 35, 'bench_press_1RM': 50, 'deadlift_1RM': 110}")
```

## 4. 평가용 데이터 로드

```python
df = pd.read_excel(f"/content/drive/MyDrive/05_[공유파일]/Burnfit_dataset/eval_data_final.xlsx")

inputs = df["input"]
instructions = df["instruction"]
```

```python
print(inputs[0])
print(instructions[0])

print(inputs[1])
print(instructions[1])

print(inputs[2])
print(instructions[2])
```

```python
outputs = []
index = 0

for input, instruction in zip(inputs, instructions):
  output = sampler.chat(f'{instruction}, {input}')
  outputs.append(output)
  print(f"index : {index} | ✅ Good")
  index += 1
```

```python
df["output"] = outputs
df.head()
```

```python
df.to_excel(f"/content/drive/MyDrive/05_[공유파일]/Burnfit_dataset/eval_data_final.xlsx", index=False)
```
